# Algorithmik und Programmieren mit Octave (bzw. MATLAB)

Behandelt werden grundlegende Konzepte der Algorithmik und des Programmierens anhand von Octave/MATLAB: Funktionen und Skripte, Vektoren und Matrizen, Kontrollstrukturen wie Schleifen und Verzweigungen, elementare Programmiertechniken wie Iteration und Rekursion und einfache grafische Visualisierungen. Die behandelten Konzepte werden jeweils anhand einfacher Beispiele eingeführt und können anschließend durch Aufgaben selbstständig ausprobiert und an zentralen mathematischen Algorithmen wie dem Sieb des Eratosthenes oder dem euklidischen Algorithmus vertieft werden.

## Lerneinheit 1: Ein erstes kleines Programm

### Lernziele

- ein einfaches Skript im Editor schreiben, das eine mathematische Berechnung durchführt, und es erfolgreich ausführen  
- den Output Ihres Programms lesen und nachvollziehen

### Ein erstes Beispiel

Kopieren Sie den folgenden Code, fügen Sie ihn in die nachstehende Zelle ein und führen Sie den Code anschließend aus.

```octave
a = 3;
b = 2
c = a + b
```

### Was passiert hier?

In diesem Beispiel werden Werte in Variablen gespeichert und anschließend verarbeitet.

### Zuweisung (=)

Wir speichern Zahlen in sogenannten **Variablen**. Das Zeichen `=` bedeutet in der Programmierung **Zuweisung**. Der Wert des Ausdrucks rechts vom Gleichheitszeichen wird in der Variable links gespeichert. Man kann sich eine Variable wie eine Box vorstellen, in die ein Wert hineingeschrieben wird.

Beispiel:
- `a = 3` bedeutet: Die Variable `a` erhält den Wert 3.
- `b` bekommt den Wert 2
- `c` bekommt die Summe von `a` und `b`

Es ist kein mathematisches "gleich", sondern eben eine Zuweisungsoperation. Alternativ wird dies manchmal auch als Pfeilnotation `a <- 3` geschrieben, um zu verdeutlichen, dass der Wert von rechts in die Variable links geschrieben wird.

- Berechnungen werden direkt beim Ausführen durchgeführt
- Die Ausführung erfolgt zeilenweise von oben nach unten
- Ohne `;` am Zeilenende wird der aktuelle Wert ausgegeben
- Mit `;` am Zeilenende wird die Ausgabe unterdrückt (aber trotzdem intern durchgeführt)

### Aufgabe:

Ändern Sie im obigen Code Beispiel die Werte von `a` und `b` und geben Sie anderen Zeilen am Ende ein Semikolon hinzu oder nehmen es weg und beobachten Sie das Ergebnis.

## Lerneinheit 2: Eigene Funktionen schreiben

Funktionen sind die Grundbausteine, um komplexe Skripte in kleine, wiederverwendbare Einheiten zu zerlegen.

---

### Lernziele

- den Aufbau einer Funktion (`function ... end`) verstehen und anwenden
